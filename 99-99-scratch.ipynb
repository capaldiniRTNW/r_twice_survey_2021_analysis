```python
def count_binary_digits(n):
    """Count the number of digits in the binary representation of n."""
    if n == 0:
        return 1  # Special case: 0 is represented as "0" (1 digit)
    return len(bin(n)) - 2  # Subtract 2 to remove the "0b" prefix
```
